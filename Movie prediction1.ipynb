```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from sklearn.ensemble import RandomForestRegressor
```

```python
df=pd.read_csv("IMDb Movies India.csv",encoding="latin")
print(df)
```

```python
print("print first 5 values")
print(df.head())

print("\nInformation")
print(df.info())

print("missing value")
print(df.isnull().sum())
```

```python
df = df.dropna(subset=["Rating"])
df
```

```python
'''Fill Missing Values'''
df["Genre"] = df["Genre"].fillna("Unknown")
df["Director"] = df["Director"].fillna("Unknown")
df["Actor 1"] = df["Actor 1"].fillna("Unknown")
df["Actor 2"] = df["Actor 2"].fillna("Unknown")
df["Actor 3"] = df["Actor 3"].fillna("Unknown")
print(df.isnull().sum())
```

```python
'''Clean Year'''
df["Year"] = df["Year"].str.extract("(\d{4})")
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["Year"] = df["Year"].fillna(df["Year"].median())
df
```

```python
'''Clean Duration'''
df["Duration"] = df["Duration"].str.replace(" min", "", regex=False)
df["Duration"] = pd.to_numeric(df["Duration"], errors="coerce")
df["Duration"] = df["Duration"].fillna(df["Duration"].median())
df
```

```python
'''Clean Votes'''
df["Votes"] = df["Votes"].str.replace(",", "", regex=False)
df["Votes"] = pd.to_numeric(df["Votes"], errors="coerce")
df["Votes"] = df["Votes"].fillna(df["Votes"].median())
df
```

```python
''' Encode Categorical Columns'''
le = LabelEncoder()

df["Genre"] = le.fit_transform(df["Genre"])
df["Director"] = le.fit_transform(df["Director"])
df["Actor 1"] = le.fit_transform(df["Actor 1"])
df["Actor 2"] = le.fit_transform(df["Actor 2"])
df["Actor 3"] = le.fit_transform(df["Actor 3"])
df
```

```python
'''Features and Target ,"Year","Duration","Votes",  '''
X = df[["Genre","Director","Actor 1","Actor 2","Actor 3"]]
y = df["Rating"]

'''Train-Test Split'''
X_train, X_test, y_train, y_test = train_test_split( X,y,test_size=0.20,random_state=42)

'''Train Model'''
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
'''Prediction'''
y_pred = model.predict(X_test)
```

```python
'''Evaluation'''
print("\nModel Performance")

print("Mean Absolute Error :", mean_absolute_error(y_test, y_pred))

print("Mean Squared Error :", mean_squared_error(y_test, y_pred))

print("R2 Score :", r2_score(y_test, y_pred))
```

```python
'''Compare Actual vs Predicted'''
result = pd.DataFrame({
    "Actual Rating": y_test,
    "Predicted Rating": y_pred
})

print("\nActual vs Predicted Ratings")

print(result.head(10))
```

```python
'''Predict One Sample Movie'''
sample = X_test.iloc[[0]]

prediction = model.predict(sample)

print("\nActual Rating :", y_test.iloc[0])
print("Predicted Rating :", prediction[0])
```
